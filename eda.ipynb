```python
import gzip
import json

path = "data2/dataset2.json.gz"

with gzip.open(path, "rt", encoding="utf-8") as f:
    for i, line in enumerate(f):
        record = json.loads(line)
        print(record)
        if i == 2:
            break
```

```python
with gzip.open("data2/dataset2.json.gz", "rt", encoding="utf-8") as f:
    record = json.loads(next(f))

transcript_id = next(iter(record))
position = next(iter(record[transcript_id]))
sequence = next(iter(record[transcript_id][position]))
reads = record[transcript_id][position][sequence]

print("transcript:", transcript_id)
print("position:", position)
print("sequence:", sequence)
print("number of reads:", len(reads))
print("first read's 9 features:", reads[0])
```
